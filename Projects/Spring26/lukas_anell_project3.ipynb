## Question 1

```python
import pandas as pd
import numpy as np


video_games = pd.read_csv("/anvil/projects/tdm/data/grouping/vgchartz-2024.csv")
video_games.head()
```

```python
video_games.groupby("genre", as_index=False)["critic_score"].mean().sort_values(by="critic_score", ascending=False).head()
```

```python
video_games.groupby("console", as_index=False)["total_sales"].sum().sort_values(by="total_sales", ascending=False).head()
```

I had to look up how to sort the values, but it wasn't too difficult in the end.

1.1: The Sandbox genre had the highest average critic_score.

1.2: The PS2 had the highest number of total_sales.

1.3: Grouping data in Python is useful because it allows you to see how certain parts of the data are related to each other. For example, we can group the data to find which game genre has the highest critic score.

```python
app_store = pd.read_csv("/anvil/projects/tdm/data/grouping/googleplaystore.csv")
app_store.head()
```

```python
app_store.dtypes
```

```python
app_store["Clean_Reviews"] = app_store["Reviews"].str.replace(",", "", regex=False)
app_store = app_store[app_store["Clean_Reviews"] != "3.0M"]
app_store["Numeric_Reviews"] = app_store["Clean_Reviews"].astype(float)
app_store["Numeric_Reviews"].head()
```

```python
app_store["rating_per_review"] = app_store["Rating"] / app_store["Numeric_Reviews"]
app_store["rating_per_review"].head()
```
