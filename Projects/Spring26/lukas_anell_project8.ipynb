```python
# song durations in seconds
ts_songs['duration_sec'] = ts_songs['duration_ms'] / 1000
# song durations in minutes
ts_songs['duration_min'] = ts_songs['duration_sec'] / 60
```

```python
total_by_album = ts_songs.groupby("album")["duration_min"].sum().reset_index()
total_by_album.head()
```

```python
long_albums = total_by_album[total_by_album["duration_min"] > 80]
long_albums.head()
```

```python
def find_albums_by_time(input_df, input_min_duration):
    total_by_album = input_df.groupby('album')['duration_min'].sum() 
    long_albums = total_by_album[total_by_album > input_min_duration] 

    return long_albums
```

```python
min_80 = find_albums_by_time(ts_songs, 80)
min_80.head()
```

```python
min_120 = find_albums_by_time(ts_songs, 120)
min_120.head()
```

```python
lower_threshold = 30
upper_threshold = 80
```

```python
def classify(duration):
        if duration < lower_threshold:
            return "short_name"
        elif duration > upper_threshold:
            return "long_name"
        else:
            return "other/default_name"
```

```python
total_by_album = ts_songs.groupby("album")["duration_min"].sum().reset_index()   
```

```python
total_by_album["duration_min"].apply(classify).head()
```

```python
def albums_by_length(df, lower_threshold, upper_threshold):
    total_by_album = df.groupby("album")["duration_min"].sum().reset_index()                
    
    def classify(duration):
        if duration < lower_threshold:
            return "short_name"
        elif duration > upper_threshold:
            return "long_name"
        else:
            return "other/default_name"
    
    total_by_album["category"] = total_by_album["duration_min"].apply(classify)
    
    return total_by_album
```

```python
test = albums_by_length(ts_songs, lower_threshold=30, upper_threshold=80)
test.head()
```

```python
youtubers['subscribers2'] = youtubers['subscribers'].str.replace(',', '').astype(float)
youtubers.head()
```

```python
def top_youtuber_by_genre(df, genre):
    genre_rows = df[df['category'] == genre]
    top_index = genre_rows['subscribers2'].idxmax()

    return genre_rows.loc[top_index]
```

```python
top_gaming = top_youtuber_by_genre(youtubers, "Gaming")
print(top_gaming)
```

```python
top_gaming = top_youtuber_by_genre(youtubers, "Music")
print(top_gaming)
```
