## Question 4

```python
import pandas as pd
my_flights = pd.read_csv('/anvil/projects/tdm/data/flights/subset/1997.csv')
```

```python
my_flights.head()
```

```python
bins = [0, 300, 600, 900, 1200, 1500, 1800, 2100, 2400]
labels = ["Late Night", "Early Morning", "Morning", "Late Morning", "Afternoon", "Early Evening", "Evening", "Night"]

my_flights["depart_times"] = pd.cut(my_flights["DepTime"], bins=bins, labels=labels)
my_flights.head()
```

```python
my_flights["depart_times"].value_counts().head()
```

```python
my_flights["arrival_times"] = pd.cut(my_flights["ArrTime"], bins=bins, labels=labels)
my_flights.head()
```

```python
my_flights["arrival_times"].value_counts().head()
```

4.1: See above.

4.2: See above.

4.3: Morning is the busiest departure time, early evening is the busiest arrival time.
